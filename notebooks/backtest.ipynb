# Backtesting

we need tools to be able to check how well our model works, both in stable and chaotic times. this is the baseline of how we stress test our model and see what needs to be improved. we'll look at many methods: simple var (checking to see how often the model is more optimistic than real life), kupiec (checking to see if the distribution of errors matches what we expect to see), christoffersen (checking to see if errors cluster or not)

### Value-at-Risk exceptions 

the main method to check whether our model is predicting realistic outcomes is to check how many times, how often, and in what way our predictions differ from reality. 

## value at risk backtesting

value at risk backtesting consists of taking the 99th percentile var and checking whether reality falls into that exception bracket (did reaklity fall beneath the percentile issue). we do this on a rolling basis over the course of n training days. we want the amount of issues to be around about 1% of the number of training days. 

## kupiec backtesting

think of exceptions as being binomially diustributed. L0 is the ideal distribution, L1 is the data suggested one, with LR being the log of the quotient of these
if L0 similar to L1, log(L0/L1) -> 0
if the model is correct then LR is distributed like chi2(1), therefore take 95% confidence interval of
that and compare with calculated LR
if 95% CI of chi2(1) is larger than LR, don't reject model as "number of exceptions is statistically consistent
with the claimed confidence level"
